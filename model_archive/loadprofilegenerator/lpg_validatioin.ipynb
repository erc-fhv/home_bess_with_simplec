# Comparison of LPG loadprofiles with standard loadprofiles

```python
year = 2021 # 2025 # 
df_synth = pd.read_excel(f'data/synthload{year}/synthload{year}.xlsx', index_col=0, header=0, skiprows=[1], parse_dates=True)
df_synth.index.inferred_freq
```

```python
fig = go.Figure()

for name, df in dfs_resampled.items():
    if 'Electricity_HH1' in df.columns:
        fig.add_trace(
            go.Scatter(
                x=df.index,
                y=df['Electricity_HH1'],
                line= {'shape': 'hvh'},
                name=name
            )
        )

fig.add_trace(
            go.Scatter(
                x=df_synth.index,
                y=df_synth['H0'] * 3.6, # multiplied by 3.6 MWh to scale to standard household load 
                line= {'shape': 'hvh'},
                name='Synthetic Load'
            )
        )

fig.update_layout(
    title="Electricity_HH1 for all households",
    xaxis_title="Time",
    yaxis_title="Electricity_HH1 [unit]",
    template="plotly_white",
    legend_title="Profile"
)

fig.show()
```

## Findings
- Even though not ploted to scale:
- The figure shows qualitative aggreement of standard load profiles and lpg profiles
- there are a clear daily and a weekly patterns visible
- Seasonal dependency can not be seen as clearly (however this seems to only be the household load without heating!)

# Conclusion:
- The house electricity seems not reasonable.
- Due to the low seasonal dependency of the heat pump building, the heat pump house should not be chosen.
- Also the gas heating house model seems not reasonable, why should a gas heating house have similar electrical loads to a heat pump house?

-> only use the household profiles, discard House Electricity. 

- The load profiles for each household have signifficantly higher peak loads than the standard load profiles.
- But the gneral weekly dependency is present.
- Also a much greater variability is present in the profiles, which seems much more realistic than the standard load profiels for the purpose of the simulation.
- Particularly for forecasting, those profiles can maybe regarded as a worst case scenario rather.
- The higher discrepancy between peak and base load will on the other hand exagerate the effect of flexibilities compared to a profile with a smoother/flatter distribution.

-> Household profiles seem like a reasonable choice for analysis regarding flexibility options and control of such.

-> A detaild investigation of internal gains an hot water consumption is still lacking and should be cunducted in the future, to integrate with building and DHWH models.
